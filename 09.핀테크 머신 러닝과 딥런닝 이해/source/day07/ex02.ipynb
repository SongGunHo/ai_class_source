```python
# 사과 파인애플  바나나 각각의 픽셀의평균
apple = fruits[0:100].reshape(-1, 100* 100) # 너비 높이 -> 1차원 배열 10000
```

```python
apple = fruits[0:100].reshape(-1, 100* 100) # 너비 높이 -> 1차원 배열 10000
pineapple = fruits[100:200].reshape(-1, 100* 100) # 너비 높이 -> 1차원 배열 10000
banana = fruits[200:300].reshape(-1, 100* 100) # 너비 높이 -> 1차원 배열 10000
```

```python
apple.shape
```

```python
np.mean(apple, axis=1) #axis: 0은 행 ㅡ  1은 열 
```

```python
np.mean(pineapple, axis=1)
```

```python
np.mean(banana, axis=1)
```

```python
#__, axs = plt.subplots(1,3, figsize=(20,10))
plt.
axs[0].hist(np.mean(apple,axis=1), alpha=0.8)
axs[1].hist(np.mean(pineapple,axis=1), alpha=0.8)
axs[1].hist(np.mean(banana,axis=1), alpha=0.8)
plt.legend(['apple','pineapple','banana'])
```

```python
# 100개 사과 평균 픽셀 100개 바나나 평균 픽셀 ,  100개 파이에플 평균 픽셀 
apple_mean = np.mean(apple, axis=0).reshape(100, 100)
pineapple_mean = np.mean(pineapple, axis=0).reshape(100,100)
banana_mean = np.mean(banana, axis=0).reshape(100,100)

_, axs = plt.subplots(1, 3, figsize=(15, 5)) # plt 는 그래프를 그려주는 역활 
axs[0].imshow(apple_mean, cmap='gray_r')
axs[1].imshow(pineapple_mean, cmap='gray_r')
axs[2].imshow(banana_mean, cmap='gray_r')


plt.show() # 그래프로 화면에서 표시하는 명령어 
```

```python
apple_diff = np.abs(fruits - apple_mean)
apple_diff_mean= np.mean(apple_diff, axis=(1,2))
apple_diff
```

```python
apple_index = np.argsort(apple_diff_mean)[:100]# 모든 사과의 평균과 차이가 가장 작은 순서대로 정렬하고 그 인덱스 번호 반환  np.argost 배열을 정렬 하는것 
apple_index
```

```python
_, axs = plt.subplots(10,10, figsize=(10,10))
for i in range(10):# 10행
    for j in range(10): # 10열 
        axs[i,j].imshow(fruits[apple_index[i* 10+j]], cmap='gray_r')
        axs[i,j].axis('off')
plt.show()
```

```python
pineapple_diff = np.abs(fruits - pineapple_mean)
pineapple_diff_mean= np.mean(pineapple_diff, axis=(1,2))
pineapple_diff
```

```python
pineapple_index = np.argsort(pineapple_diff_mean)[:100]# 모든 사과의 평균과 차이가 가장 작은 순서대로 정렬하고 그 인덱스 번호 반환  np.argost 배열을 정렬 하는것 
pineapple_index
```

```python
_, axs = plt.subplots(10,10, figsize=(10,10))
for i in range(10):# 10행
    for j in range(10): # 10열 
        axs[i,j].imshow(fruits[pineapple_index[i* 10+j]], cmap='gray_r')
        axs[i,j].axis('off')
plt.show()
```

```python
banana_index = np.argsort(banana_diff_mean)[:100]# 모든 사과의 평균과 차이가 가장 작은 순서대로 정렬하고 그 인덱스 번호 반환  np.argost 배열을 정렬 하는것 
banana_index
```
