```python
from scipy.stats import randint, uniform
```

```python
ri = randint(0, 10)# 0~ 9 무작위 수 
ri.rvs(10)# 0~9 숫자 중 무작위 수를 10개 생성 
```

```python
import numpy as np
np.unique(ri.rvs(1000), return_counts=True) # 증복된 총 갯수를 함께 반환 
```

```python
ui  = uniform(0, 1) # 0~1 사이의 무작위 실수 
ui.rvs(5)# 0~1 사이에 무작위 실수를 5개 생성
```
