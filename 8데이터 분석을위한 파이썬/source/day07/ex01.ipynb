```python
import time
import threading
```

```python
def some_work():
     for i in range(1,11):
        print("작업 %d" %i)    
        time.sleep(1)

def some_work2():
     for i in range(1,6):
        print("작업 %d" %i)
        time.sleep(1)
```

```python
th1= threading.Thread(target=some_work) # 쓰레드 생성 
th1.start()# 호출 스택 + 실행
th2= threading.Thread(target=some_work2)
th2.start()


th1.join()
th2.join()

for i in range(1, 11):
    print("메인 작업 %d " %i)
    time.sleep(1)
```
