# 자주 사용하는 문자 클래스 
-` \\d` : \[0-9\] : 숫자   
- `\\D` :\[^0-9\] :  숫자가 아닌 패턴  
- `\s`: 공백 문자 한개(`'', \t -탭 한칸 . \r , \n , \f`) - [\s\t\r\f\n] 
- `\s` : 공백이 아닌 모든 문자 
- `\w` : \[a-zA-Z0-9_\] :단어 한개 패턴 참고로 한글 간어 한개 [`a-zA-Z0-9_ㄱ-ㅎ가-힣`]
- `\W` : \[^a-zA-Z0-9_\] 참고로 한글 간어 한개



# . 문자 : 문자 1개 
- `\n`을 제외한 문자 1개 (줄 개행 문자는 포함 x)

````python
# a0b, a1b, a2b, a!b 
words= ['a0b', 'a1b', 'a2b', 'a!b ']

p=re.compile("a.b")
for word in words:
    print(word, p.match(word))
    print("\n")
````

````python
words= ['a000b', 'a111b', 'a222b', 'a!b' ]

p=re.compile("a...b")
for word in words:
    print(word, p.match(word))
````

# 패턴의 반복 횟수
- 패턴 {숫자 } : 패턴이 숫자 만큼 반복 
- 패턴{시작 번호 , 종료 번호 } : 패턴이 시작 번호 이상 종료 번호 이하 반복 
    - 패턴 {1,3}: 패턴이 1이상 3번 이하 
    - 패턴 
- 패턴{시작 번호 , } : 패펀이 시작 번호 이상 반봅
    - 패턴{1 ,} : 패턴이 1번이상 반복  / 패턴 +
    - 패턴{0, }: 패턴이 0번 이상 반복 , 패턴에 해당 하는 무자가 있어도 도히고 없어도 된다  /패턴 *

````python
words= ['a000b', 'a111b', 'a222b', 'a!b' ,'a3333333b' , 'ab']

#p=re.compile("a.{3}b")
#p = re.compile("a.{1, 3}b")

#p = re.compile("a.{1,}b")
#p = re.compile("a.{0,}b")
#p = re.compile("a.+b")# a.{1,}b /.이 한개 이상 
p= re.compile("a.*b" ) # a.{0,}b /. 이 0개 이상 
for word in words:
    print(word, p.match(word))
````

## 휴대폰 번호 패턴 
- 010 -100-2000

````python
p= re.compile("010.?\\d{4}.?\\d{4}")
````

````python
p.search("010-2253-3951")
````

### ^, $
- ^ 패턴 : 패턴으로 시작 하는 패턴 
- $ 패턴 : 패턴으로 끝나는 패턴 

````python
p= re.compile("^010.?\\D{4}.?\\D{4}$") # 010으로 시작 하는 패턴 
````

````python
p.search("010-2253-3951")
````
