## 이터레이터 만들기
- 클래스에 `__iter__` `__next__` 를 정의하고 구한다 
- 커서가 존재 `__next` 호출될떄 커서를 이동 바복을 멈추는 조건 처리 꺼내진 요소 반환 
- `next` 를 호출 하면 `__next__` 호출됨

```python
fruits = ['사과', '오렌지' , '멜론' , '바나나' , '망고']
```

```python
itr = iter(fruits)
type(itr)
```

```python
next(itr)
```

```python
itr = iter(fruits)
while True:
    try:
        print(next(itr))
    except StopIerastion:
        break    
```

## 제너레이터 
- 제너레이터란 
    - 이터레이터 로 만들어주는 함수 
    - 실행 밎 반환값 지연 return  대신 yuield 사용(여러개 사용 ) 
- 제네레이터 표현식 
    - 튜플에다가 리스트 컴프리헨션과 비슷하게 

```python
def mygen ():
    yield 'A'
    yield 'B'
    yield 'C'
    yield 'D'
```

```python
gen = mygen()
type(gen)
```

```python
next(gen)
```
