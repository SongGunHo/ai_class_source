##  파일을 읽고 쓰기
- 파일을 작업을 하려면 파일을 열어준다
```python
    open('파일경로', "모드")
```
- 모드
    - r : 읽기
    - w : 쓰기
    - a : append 약자 데이터 끝에서 누적 방식으로 추가 
    - b : r,w,a 와 함께 쓰이며 바이너리 데이더틀 작업 할때      rb,  wb
- 메모리에 자원 할당 하므로 작업을 완료후 해제가 필요하다 
```python
    f.close()
```
- 파일 작성 
```python
    f.write("데이터")
```
- 파일 읽기
```python
    f.read("")# 데이터 전체 읽기
```
