## 모듈
-  모듈이란?
1) 모듈이란 함수나 변수 또는 클래스를 모아 놓은 파이썬 파일이다.
2) 모듈은 다른 파이썬 프로그램에서 불러와 사용할 수 있도록 만든 파이썬 파일이라고도 할 수 있다.

- 모듈 만들기
- 모듈 불러오기 
``` python
import 모듈명
from 모듈 이름 포함한 패키지 경로 import  함수 또는 변수 클래스명 
from 모듈 이름 포함한 패키지 경로  import *  : 해당 모듈의 모든 함수 변수 클래스 등등 
```
> 모듈명은 확장자를 제외한 파일 명 
    - 모듈 사용하기 
    ``` python
    모듈명.함수명(..)
    모듈명.변수명  
    ```
- 모듈이 같은 폴더 안에 있을때는 import 로 바로 불러 올 수 있음
    - 같은 폴더에 없으면 불러 올 수 없다 

````python
import mod1
````

````python
mod1.add(10, 20)
````

````python
mod1.sub(20, 10)
````

````python
from  mod1 import add , sub 
````

````python
add(10, 20)
````

````python
sub(20 , 10)
````

````python
from mod1 import *
````

````python
add(10,20)
````
