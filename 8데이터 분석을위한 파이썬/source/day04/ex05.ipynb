- 다중 상속이 가능 
- 하위 클래스에서 상위 클래스에 정의된 동일한 메서드를 정의해서 기능을 변경 하는 것 : 메서드 재정의 

````python
class MoreCalculator(Calculator):
    def div(self): # 메서드 재정의
        if self.b == 0:
            return 0
        return self.a / self.b     
    
````
