##  생성자 메서드
```python
    Class ...
        def __init__(self, 메개 변수 정의)
            ....
    ...
    - 인스턴스 변수의 초기화 
    -  객체 생성시 초기화 할 값을 직접 대입 
``` 

````python
class Calculator:
    def __init__(self,a , b):
       self.a = a
       self.b = b 

    add(self):
        return self.a +self.b
````

````python
cal1 =Calculator
````

````python
#cal1.a =100
#cal2.b =200
cal1.setData(100, 200)
cal1.add()
````

````python
cal1 = Calculator # 객체 생성 
````

````python
cal2  = Calculator 
````
