## 클래스 
``` python
    class ...
        def __call__(self, 매게 변수 )
            ..
```
- 객체 참조 변수를 함수 처럼 호출 하면 '__call__'  함수 호출

````python
class Accumukator:
    def __init__(self):
        self.num = 0
    def __call__(self, num):
        self.num += num

        return self.num
````

````python
acc  = Accumukator
````

````python
acc(10)
````
