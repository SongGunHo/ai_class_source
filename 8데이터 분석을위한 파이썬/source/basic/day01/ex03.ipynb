## 문자열 인덱싱
- 각 문자 접근은 인덱스 연산자 \[\] 를 통해서 접근 
## 문자열 슬라이싱
- a [시작 번호: 종료번호 ] -> 시작 번호 \<= a \< 종료 번호 
-a\[시작 번호] -> 시작 번호부터 끝까지
-a[: 종료 번호 ] -> 종료 번호 미만 문자열을 모두 추출 
-인덱스 번호가 음수 , 오른쪽 부터 위치를 계산
    --1 오른쪽 첫번째 문자
-a\[:]     
> id\(..\) :자원의 주소값을 확인

## 문자열 형식화
-%d
-%f:실수
    -%10.4f -> 전체 자리수는 10자(정수 부분 + 소수점 자리 )에서 소수점은 4자까지만 출력 
    -%.4f -> 소수점 자리수만 4자리로 한정 (자바 x)
    -%0.4f ->(자바 o)
-%s
    -%10s -> 왼쪽부터 최소10자 , 10자가 안되는 경우 왼쪽부터 공백으로 채워준다 
    -%-10s -> 오른쪽 부터 최소 10자 , 10자 안되는 경우 오른쪽부터 공백이 생긴다 
## 적용방법
- "형식화된 문자열 " % 대체될 값
- "형식화될 문자열 (대체될 부분이 여러개 인경우)"%(값1, 값2, ...)
- 형식화될 문자열 (대체될 부분이 여러개 인경우)"%값1, 값2
- 형식화 할때 % 를 문자열로 상용하는  경우 %% 로 대채
- 문자열 객체의함중에서 format 함수 사용 
    - 형식화 할때 치환 될 부분을 {0}. {1}로 순서 번호 형태로 지정 
    - 형식화 할때 치환 될 부분을 직접 이름형태로 지정 할 수도 잇음 {mumber} ,{day}
    - 숫자와이름을 혼용해서 형식화 문자열을 작성
    - {0:\< 10} : 최소 10자리 남는 자리수를 오른쪽 부터 공백
        - {0:\_<10}:최소 10자리, 남는 자리수을 오른쪽 부터 _로 채워 넣는다
    - {0:\ >10} : 최소 10자리 남는 자리수를 외쪽 부터 공백
        - {0:\_!>10}:최소 10자리, 남는 자리수을 왼쪽 부터 _로 채워 넣는다
    - {0:^10 }: 최소 10자리 남은 자리수는 좌우 균등하게 분배
        - {0:\0^10}:최소 10자리, 남는 자리수을 균등하게 0 으로 채워 넣는다
    - {0:104f} :최소 10자리 (정수부 + 소수점 )를 유지 소수점은 4자리로 한정 
        - {0:4f}: 소수점만 4자로 한정
    
- 문자열 앞에 "f", 파이썬 3.6부터 추가 변수 그 자체르 ㄹ문자열에 추가

## f 형식화
- 문자열 앞에 f를 붙이면 된다 
- 변수를 직접 형식화 문자열에 출력
- 간단한 연산 , 함수 호출도 가능 
- {값 :\< 10} -> 최소 10자리 , 남은 자리수는 오른쪽 공백 
- {값 !:\ 10 } -> 최소 10자리 남은 자리수는 오른쪽으로 ! 채워 넣은다 

- {값:\> 10 } -> 최소 10자리 남은 자리수는 왼쪽으로 공백
- {값:^10} -> 최소 10자리 남은 자리수는 좌우 균등하게 공백
- {실수 :10.4f}-> 정수 + 소수점 자리가 최소 10자리 , 소수점은 4자리 한정 
- {실수 :4f} -> 소수점은 4자리 한정
- {} 사용 가능 

```python
name ="이이름"
age =40
num = 10
```

```python
f'이름:{name}, 나이: {age} , 숫자:{num}' #문자열 앞에 f를 붙이면 된다 
```

```python
f'이름:{name:<10}, 나이: {age:>10} , 숫자:{num:^10}'
```

```python
f'이름:{name:!<10}, 나이: {age:*>10} , 숫자:{num:=^10}'
```

```python
 num = 100.12345678 # {실수 :10.4f}-> 정수 + 소수점 자리가 최소 10자리 , 소수점은 4자리 한정 
 f'{num: 10.4f}'
```

```python

 f'{num:.4f}' #{실수 :4f} -> 소수점은 4자리 한정
```

```python
f'{{값}}' #{} 사용 가능 
```

## 문자열 관련 함수들
- 문자 개수 세기 count
- 문자 개수 세기 - count
        1. 문장에서 특정 문자가 몇번 등장 하는지 
-  위치 알려 주기 1 - find
        1. 지정된 문자를 못 찾은 경우→ 값을 반환
- 위치 알려 주기 2 - index
- 특징 문자열로 결합 하여 문자열 생성 - join
    - 문자열 외에 list tuple 사용 가능 
    - 대문자 -> 소문자 -> upper
- 공백 제거 - strip / 좌 , 우 공백 제거
      - lstrip : 왼쪽 여백 제거
      - rstrip : 오른쪽 여백 제거
- 문자열 치환 : replace
- 문자열 나누기 : split
        - 문자열 분리 구분문자를 지정하지 않느면 공백 기준에서 분리 

```python
 a = "One Little, Two Little, Three Little Indians"
```
