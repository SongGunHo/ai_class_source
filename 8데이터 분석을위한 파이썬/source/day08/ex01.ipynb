    1. import re(”패턴  re 옵션 상수)
    2. DOTALL, S
        1. re.DOTALL 
            1. .는 기본 동작으로\N 제외한 모든 문자이지만 re.DOTALL 컴파일 옵션을 추가하면 \\n도 포함한 모증 문자 1게로 인식 
    3. IGNORECASE, I 
        re.IGNORECASE : 대소문자 구분없는 패턴 /re.I
    4.  MULTILINE, M
        - 일반 패턴은 한줄만 패턴을 체크
        - 여러줄에 걸친 한줄 한줄 패턴을 체크 
        - re.MULTILINE / re.M
    5. VERBOSE, X
        - 패턴의 해석에 대한 자세한 정보 re.VERBOSE / re.X
    6. 여러개 컴파일 옵션을 사용 하는 겨우 | 연산자 로 이어 준다 예) re.I , re.M 

````python
import re
words = ["a11234b", "a   b", "a!b" , "a\nb"]
````

````python
#p = re.complie("a.*b")
#p = re.compile("a.*b" , re.DOTALL) # . - 줄개행 문자 (\n)을 포함한 모든 문자로 1개 인식
p= re.compile("a.*b", re.S) 
for word in words:
    m = p.match(word)
    print(m)
````

- \\A: 여러줄로 구성되 문장에서 패턴으로 시작 
- 패턴\\Z: 여러줄로 수성된 문장에서 패턴으로 종료

````python
p.re.compile(r"^python.+\d$")
m = p.match(str)
````
