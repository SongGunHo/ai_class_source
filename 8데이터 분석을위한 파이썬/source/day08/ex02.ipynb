2. ^
    - : 패턴 시작 하는
3. $
    - $ : 패턴으로 끝나는 문장 
4. \A
    - 여려줄 전체 문장에서 시작 하는 문장
5. \Z
    - 여러줄 전체 문장에서 끝나는 문장 
6. \b
    - 문자 클래스 내에서 사용 되는 \[\\b] : 백스프페이스 키를 한번 누른것 (한 글자 귀에서 부터 삭제 )
    - 문자 클래스가 아닌 패턴으로 사용되는 경우 : 단어의 경계선을 의미 한다 (공백 줄 갸향 문자등 확인 한다 )
7. \B
    - \[^\\b\]
    -  단어의 경계선이 아닌 문자
