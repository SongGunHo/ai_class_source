````python
import re
````

````python
p= re.compile(r"(ABC){1,3}")
m = p.search(str)
m
````

````python
str = """ 이이름 010-111-2222
김이름 010-333-4444
Rechel 010-3000-3000"""
````

````python
#p = re.compile(r"^[a-zㄱ-ㅎ-가-힣]+\s+010-\D*\d{4}\D*\d{4}$", re.M |re.I)
p = re.compile(r"(^[a-zㄱ-ㅎ-가-힣]+)\s+(010-\D*\d{4}\D*\d{4}$)", re.M |re.I)
#p2 = re.compile(r"^.")
#m = p.search(str)
````

````python
items = p.findall(str)
items
````

````python
for m in p.findall(str):
    print(f"m.group():{m.group()}")
    print(f"m.group(1):{m.group(1)}")
    print(f"m.group(2):{m.group(2)}")
````

## 문자열 바꾸기

1.  sub 
    - count : 지정 하지 않으면 전체 문자열에서 모든 패턴을 치환 숫자를 지정 하면 지정괸 숫자 만큼 패턴 치환

2. subn
    -sub + count 기능 포함 반환값은 튜플 

````python
p = re.compile(r"(^[a-zㄱ-ㅎ-가-힣]+)\s+(010-\D*\d{4}\D*\d{4}$)", re.M |re.I)
````

````python
str2 = p.sub(r"\g<2> \g<1>", str)
print(str2)
````

- 그룹핑의 이름 
```
(?P<이름>)
```

````python
p = re.compile(r"(?P<name>^[a-zㄱ-ㅎ-가-힣]+)\s+(P<moblie>010-\D*\d{4}\D*\d{4}$)", re.M |re.I)
````

````python
str2 = p.sub(r"\ g<mobile> \g<name>" ,str)
print(str2)
````

## 전방 탐색

1. 긍정형 전방 탐색
    - 특정 패턴의 앞쪽 패턴으로 한정 
    ```
    앞쪽 패턴 (?=특정 패턴 )
    ```
2. 부정형 전방 탐색
    - 특정 패턴이 아닌 앞쪽 패턴으로 진행 
    ```
    앞쪽 패턴 (?! 특정패턴 )
    ```

````python
str = """ http://www.naver.com
http://www.daum.net
http://www.webnmobile.net
"""
````

````python
#p = re.compile(r"http[s]?://", re.M)
p = re.compile(r"http[s]?(?=://)")
````

````python
items = p.findall(str)
items
````

````python
p= re.compile(r"http[s]?://(www\.)?({^.}+(?=\.net$))", re.M) # 전방 탐색
````

````python
for item in p.finditer(str):
    print(item,group(2)) # 전방 부정 탐색 
````

````python
p= re.compile(r"http[s]?://(www\.)?({^.}+(?!=\.net$))", re.M) ## 전방 부정 탐색 , net
````

### greedy와 non-greedy
- 패턴 + , 패턴* , 패턴? 패턴 {n, m} : 최대 매칭(greedy)
- 패턴+? 패턴*? 패턴?? 패턴 {n,m }: 최소 매칭 (non -greedy)

````python
str ="<html><head><title>사이트 제목</title></head><body></body></html>"
````

````python
p = re.compile(r"<.+>")
````

````python
m = p.search(str)
m.group()
````

````python
p = re.compile(r"<.*?>") # non-greedy최소 매칭 
m = p.search(str)
m.group()
````
