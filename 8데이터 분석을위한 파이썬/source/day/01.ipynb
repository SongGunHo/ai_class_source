## if  구조
``` python
    if 조건식 :
        참일떄 실행 되는 코드
    
    if 조건식 : 
        참일때 실행 되는 코드
    else :
        거짓일때 실행 되는 코드 \
    
    
    if 조건식 1:
        조건식 1 참일떄 실 행되는 코드
    else if 조건식 2:
        조건식이 1 이 거짓일때 조건식 2 참임     

    else : 
        거짓일 떼 실행 되는 코드


    if 조건식 :  실행 코드 
```



````python
age  = 13
if age >= 14:
    pass # 아무것도 하지 않음 
else :
    print("가입 불가능 ")    
````
