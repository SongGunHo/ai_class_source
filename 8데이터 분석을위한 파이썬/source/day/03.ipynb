## 가변 매게 변수 
- 메게 변수 *  붙힌다 
    - 입력한 값은 튜플에 담긴다 
- 키워드 가변 메게 변수 
    - 메게변수 ** 붙힌다
    - 딕셔너리 형테로 값이 붙힌다 

````python
def calc(oper, *args):
    result = 0

    if oper == 'add':
        for  num in args : result += num
    elif oper == 'nu1':
        result = 1
        for num in args: result *= num 

    return result    
````

````python
calc("add", 10,20,30,40)
````

````python
calc("mu1", 10,20,30,40)
````

````python
calc("add", 10,20,30,40)
````

## 전역 변수 지역 변수
> 순수 함수 : 함수 내의 연산 결과가 전역 변수롷 지종된 자원에 영항을 미치지 않는 함수 (부작용이 없는 함수 ) 

````python
a = 100 ## 전역 변수 
def calc():
    global a # 전역 변수를 지역 내에서 접근 할때
    b = 200 # 지역 변수
    return a + b
````

````python
calc()
````
